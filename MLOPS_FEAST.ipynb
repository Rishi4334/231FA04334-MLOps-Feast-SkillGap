```python
# Clean Gender values
df["Gender"] = df["Gender"].astype("string").str.strip().str.lower()

# Encode Gender
df["gender_encoded"] = df["Gender"].map({
    "male": 0,
    "female": 1
})

# Check for values that could not be encoded
print("Missing encoded Gender values:",
      df["gender_encoded"].isna().sum())

print(df[["Gender", "gender_encoded"]].head(20))
```

```python
import pandas as pd

# ==========================================
# 1. Entity / Student ID
# ==========================================

df["student_id"] = df["Student_ID"].astype(str)


# ==========================================
# 2. Handle missing numerical values
# ==========================================

numerical_columns = [
    "Age",
    "Current_Year",
    "Semester",
    "CGPA",
    "Attendance_Percentage",
    "Internal_Marks",
    "External_Marks",
    "Mini_Project_Score",
    "Major_Project_Score",
    "Lab_Performance",
    "Assignment_Average",
    "Backlogs",
    "Academic_Consistency",
    "Class_Rank"
]

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())


# ==========================================
# 3. Encode Gender
# ==========================================

df["Gender"] = (
    df["Gender"]
    .astype("string")
    .str.strip()
    .str.lower()
)

df["gender_encoded"] = df["Gender"].map({
    "male": 0,
    "female": 1,
    "other": 2
}).astype("int64")


# ==========================================
# 4. Derived Feature 1
# Academic Performance Score
# ==========================================

df["academic_score"] = (
    df["CGPA"] +
    df["Attendance_Percentage"] / 10 +
    df["Internal_Marks"] / 10 +
    df["External_Marks"] / 10
) / 4


# ==========================================
# 5. Derived Feature 2
# Project Performance Score
# ==========================================

df["project_score"] = (
    df["Mini_Project_Score"] +
    df["Major_Project_Score"] +
    df["Lab_Performance"]
) / 3


# ==========================================
# 6. Derived Feature 3
# Technical Skill Score
# ==========================================

technical_features = [
    "C_Programming",
    "C++_Programming",
    "Java",
    "Python",
    "JavaScript",
    "SQL",
    "HTML_CSS",
    "ReactJS",
    "NodeJS",
    "Programming_Logic",
    "Arrays",
    "Linked_Lists",
    "Stacks_Queues",
    "Trees",
    "Graphs",
    "Dynamic_Programming",
    "Searching_Sorting",
    "Competitive_Coding",
    "Machine_Learning",
    "Data_Preprocessing",
    "Feature_Engineering",
    "Model_Evaluation"
]

df["technical_skill_score"] = df[technical_features].mean(axis=1)


# ==========================================
# 7. Derived Feature 4
# Soft Skill Score
# ==========================================

soft_skill_features = [
    "Communication",
    "Teamwork",
    "Leadership",
    "Problem_Solving",
    "Critical_Thinking",
    "Time_Management",
    "Adaptability",
    "Creativity",
    "Presentation_Skills",
    "English_Proficiency"
]

df["soft_skill_score"] = df[soft_skill_features].mean(axis=1)


# ==========================================
# 8. Derived Feature 5
# Industry Readiness Score
# ==========================================

df["industry_readiness_score"] = (
    df["Internship_Count"] +
    df["Projects_Completed"] +
    df["Open_Source_Contributions"] +
    df["Hackathons_Attended"]
)


# ==========================================
# 9. Rename / Cast important features
# ==========================================

df["age"] = df["Age"].astype("float32")
df["cgpa"] = df["CGPA"].astype("float32")
df["attendance"] = df["Attendance_Percentage"].astype("float32")
df["backlogs"] = df["Backlogs"].astype("int64")
df["internships"] = df["Internship_Count"].astype("int64")
df["projects"] = df["Projects_Completed"].astype("int64")


# ==========================================
# 10. Target
# ==========================================

df["skill_gap_level"] = df["Skill_Gap_Level"]


# ==========================================
# 11. Check result
# ==========================================

print(df.head())
print("\nDataset shape:", df.shape)
print("\nTarget distribution:")
print(df["skill_gap_level"].value_counts(dropna=False))
```

```python
feature_definition = '''
from datetime import timedelta

from feast import (
    Entity,
    FeatureView,
    FeatureService,
    Field,
    FileSource
)

from feast.types import (
    Float32,
    Int64
)

# -----------------------------
# ENTITY
# -----------------------------

student = Entity(
    name="student",
    join_keys=["student_id"],
    description="CSE student for employability skill gap analysis"
)

# -----------------------------
# DATA SOURCE
# -----------------------------

cse_employability_source = FileSource(
    name="cse_employability_source",
    path="data/cse_employability_features.parquet",
    timestamp_field="event_timestamp",
    created_timestamp_column="created_timestamp"
)

# -----------------------------
# FEATURE VIEW
# -----------------------------

cse_employability_feature_view = FeatureView(
    name="cse_employability_features",
    entities=[student],

    ttl=timedelta(days=50000),

    schema=[
        Field(name="gender_encoded", dtype=Int64),
        Field(name="age", dtype=Float32),
        Field(name="cgpa", dtype=Float32),
        Field(name="attendance", dtype=Float32),
        Field(name="academic_score", dtype=Float32),
        Field(name="project_score", dtype=Float32),
        Field(name="technical_skill_score", dtype=Float32),
        Field(name="soft_skill_score", dtype=Float32),
        Field(name="industry_readiness_score", dtype=Float32),
        Field(name="internships", dtype=Int64),
        Field(name="projects", dtype=Int64),
    ],

    source=cse_employability_source,

    online=True
)

# -----------------------------
# FEATURE SERVICE
# -----------------------------

cse_employability_feature_service = FeatureService(
    name="cse_employability_service",
    features=[
        cse_employability_feature_view
    ]
)
'''

with open(
    f"{repo_path}/features.py",
    "w"
) as f:
    f.write(feature_definition)
```

```python
from feast import FeatureStore

store = FeatureStore(
    repo_path="/content/cse_employability_feast"
)
```

```python
feature_service = store.get_feature_service(
    "cse_employability_service"
)
```

```python
entity_df = label_df.copy()

display(entity_df.head())
```

```python
training_data = store.get_historical_features(
    entity_df=entity_df,
    features=feature_service
).to_df()
```

```python
display(training_data.head())
```

```python
feature_columns = [
    "gender_encoded",
    "age",
    "cgpa",
    "attendance",
    "academic_score",
    "project_score",
    "technical_skill_score",
    "soft_skill_score",
    "industry_readiness_score",
    "internships",
    "projects"
]
```

```python
X = training_data[feature_columns]

y = training_data["skill_gap_level"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

model.fit(
    X_train,
    y_train
)

print("Model training completed successfully.")
```

```python
predictions = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    predictions
)

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```
